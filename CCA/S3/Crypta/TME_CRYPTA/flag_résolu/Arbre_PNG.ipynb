```python
from hashlib import sha256
from cryptography.hazmat.primitives.ciphers import Cipher, algorithms, modes
from cryptography.hazmat.backends import default_backend
from cryptography.hazmat.primitives import padding
from binascii import unhexlify
```

```python
seed = '00112233445566778899aabbccddefee'
```

```python
sha256(seed.encode()).hexdigest()[16:32]
```

```python
def arbre_png(seed,length):
    A = [None]*length*2
    K=[None]*length
    X=[None]*length
    Y=[None]*length
    A[1] = bytes.fromhex(seed)
    for i in range(1,length):
        sha256_hash = sha256(A[i]).digest()
        A[2*i] = sha256_hash[:16]
        A[2*i + 1] = sha256_hash[16:32]
    
    for i in range(0,length):
        K[i]=A[i+1024]
        key = K[i]
        plaintext1 = bytes.fromhex('00000000000000000000000000000000')
        plaintext2 = bytes.fromhex('ffffffffffffffffffffffffffffffff')
        iv = unhexlify('00000000000000000000000000000000')
        backend1 = default_backend()
        backend2 = default_backend()
        cipher1 = Cipher(algorithms.AES(key), modes.CBC(iv), backend=backend1)
        encryptor1 = cipher1.encryptor()
        cipher2 = Cipher(algorithms.AES(key), modes.CBC(iv), backend=backend2)
        encryptor2 = cipher2.encryptor()
        # 对数据进行填充
        padder1 = padding.PKCS7(128).padder()
        padded_data1 = padder1.update(plaintext1) + padder1.finalize()
        ciphertext1 = encryptor1.update(padded_data1) + encryptor1.finalize()
        X[i]=ciphertext1[:16].hex()
        # 对数据进行填充
        padder2 = padding.PKCS7(128).padder()
        padded_data2 = padder2.update(plaintext2) + padder2.finalize()
        # 加密数据
        ciphertext2 = encryptor2.update(padded_data2) + encryptor2.finalize()
        
        Y[i]=ciphertext2[:16].hex()
    message = b''.join(bytes.fromhex(a + b) for a, b in zip(X, Y))
    commitement = sha256(message).hexdigest()
    
    return A,X,Y,commitement
        
    
```
