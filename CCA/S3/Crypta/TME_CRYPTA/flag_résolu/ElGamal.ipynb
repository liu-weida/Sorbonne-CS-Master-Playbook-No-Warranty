```python
def sha256_hash(input_string):
    # 创建一个 SHA-256 哈希对象
    sha256_hash_obj = sha256()

    # 更新哈希对象的输入数据
    sha256_hash_obj.update(input_string.encode('utf-8'))

    # 获取十六进制表示的哈希值
    hash_result = sha256_hash_obj.hexdigest()

    return hash_result

# 要加密的字符串
P= "PIERRE"
C="CISEAUX"
F="FEUILLE"
# 计算 SHA-256 哈希值
hashed_P = sha256_hash(P)
hashed_C = sha256_hash(C)
hashed_F = sha256_hash(F)
# 输出结果
print("Original String:", P)
print("SHA-256 Hash:", hashed_P)
print("Original String:", C)
print("SHA-256 Hash:", hashed_C)
print("Original String:", F)
print("SHA-256 Hash:", hashed_F)
```
